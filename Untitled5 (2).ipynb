```python
from concurrent.futures import ThreadPoolExecutor

def fetch_books_concurrently(query, max_pages=10, limit_per_page=100):
    """
    Fetch book data concurrently using multithreading
    """
    all_books = []

    def fetch_page(page):
        url = f"https://openlibrary.org/search.json?q={requests.utils.quote(query)}&page={page}&limit={limit_per_page}"
        data = fetch_books_with_retry(url)
        if data:
            return data.get('docs', [])
        return []

    with ThreadPoolExecutor(max_workers=5) as executor:
        futures = [executor.submit(fetch_page, page) for page in range(1, max_pages + 1)]

        for future in futures:
            docs = future.result()
            for doc in docs:
                all_books.append({
                    'title': doc.get('title', 'N/A'),
                    'author': ', '.join(doc.get('author_name', ['N/A'])),
                    'first_publish_year': doc.get('first_publish_year', 'N/A'),
                    'isbn': doc.get('isbn', ['N/A'])[0] if 'isbn' in doc else 'N/A',
                    'cover_id': doc.get('cover_i', 'N/A'),
                    'subjects': ', '.join(doc.get('subject', ['N/A']))
                })

    return all_books
```

```python
import requests
import pandas as pd
import time

def fetch_books_with_retry(url, retries=3, delay=2):
    for attempt in range(retries):
        try:
            response = requests.get(url, timeout=10)
            if response.status_code == 200:
                return response.json()
        except requests.exceptions.RequestException as e:
            print(f"Error fetching {url}: {e}. Retrying in {delay} seconds...")
            time.sleep(delay)
    print(f"Failed to fetch data after {retries} attempts.")
    return None

def main():
    query = "fiction" 
    max_pages = 10 
    limit_per_page = 100  
    books = fetch_books_concurrently(query, max_pages, limit_per_page)

    df = pd.DataFrame(books)
    print(df)

    df.to_csv('open_library_books_10_pages_with_subjects.csv', index=False)
    print("Book data saved to 'open_library_books_10_pages_with_subjects.csv'")

if __name__ == "__main__":
    main()
```

```python
import requests
import pandas as pd
import time
from lxml import html

def fetch_goodreads_rating(title, author):

    search_query = f"{title} {author}".replace(" ", "+")
    search_url = f"https://www.goodreads.com/search?q={search_query}"

    try:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
        }
        response = requests.get(search_url, headers=headers, timeout=10)
        response.raise_for_status()
        tree = html.fromstring(response.content)

        rating_xpath = '//span[contains(@class, "minirating")]/text()'
        ratings_text = tree.xpath(rating_xpath)

        if ratings_text:
            
            rating_info = ratings_text[0].strip()
            rating = rating_info.split(" — ")[0].split(" avg rating ")[-1]
            num_ratings = rating_info.split(" — ")[-1].split(" ratings")[0].replace(",", "")
            return rating, num_ratings
        else:
            return "N/A", "N/A"

    except Exception as e:
        print(f"Error fetching rating for {title}: {e}")
        return "N/A", "N/A"


def enrich_books_with_goodreads(df):
 
    ratings = []
    num_ratings = []

    for idx, row in df.iterrows():
        title = row['title']
        author = row['author']
        rating, count = fetch_goodreads_rating(title, author)
        ratings.append(rating)
        num_ratings.append(count)

        
        print(f"Processed: {title} | Rating: {rating} | Ratings Count: {count}")
        time.sleep(1)  

    df['goodreads_rating'] = ratings
    df['ratings_count'] = num_ratings

    return df


def main():
    
    file_name = "open_library_books_10_pages_with_subjects.csv"
    df = pd.read_csv(file_name)

    print(f"Loaded {len(df)} books from {file_name}.")

    
    enriched_df = enrich_books_with_goodreads(df)

    
    output_file = "books_with_goodreads_ratings.csv"
    enriched_df.to_csv(output_file, index=False)
    print(f"Enriched data saved to '{output_file}'")


if __name__ == "__main__":
    main()
```

```python
if 'subjects' in df.columns and 'first_publish_year' in df.columns:

    df_unique = df.drop_duplicates(subset='isbn')

    df_unique['first_publish_year'] = pd.to_numeric(df_unique['first_publish_year'], errors='coerce')
    bins = [0, 1980, 1990, 2000, 2010, 2020, float('inf')]
    labels = ['1980 or older', '1981-1990', '1991-2000', '2001-2010', '2011-2020', '2021 or newer']
    df_unique['year_range'] = pd.cut(df_unique['first_publish_year'], bins=bins, labels=labels)

    df_unique['subjects'] = df_unique['subjects'].str.split(', ')

    exploded_subjects = df_unique.explode('subjects')

    exploded_subjects['subjects'] = exploded_subjects['subjects'].str.strip().str.lower()
    exploded_subjects = exploded_subjects[~exploded_subjects['subjects'].isin(['large type books'])]
    popular_genres_by_range = exploded_subjects.groupby(['year_range', 'subjects']).size().reset_index(name='count')

    popular_genre_per_range = popular_genres_by_range.loc[
        popular_genres_by_range.groupby('year_range')['count'].idxmax()
    ]


    print("Most Popular Genre by Year Range:")
    print(popular_genre_per_range)


    popular_genre_per_range.to_csv("popular_genre_by_year_range.csv", index=False)

else:
    print("The 'subjects' or 'first_publish_year' column is missing in the dataset.")
```
